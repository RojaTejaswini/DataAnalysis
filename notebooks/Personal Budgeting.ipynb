```python
Q1, Q3 = df["amount"].quantile([0.25, 0.75])
IQR = Q3 - Q1
df = df[(df["amount"] >= Q1 - 1.5*IQR) & (df["amount"] <= Q3 + 1.5*IQR)]
```

```python
df["category"] = df["category"].replace("_", " ", regex=True)
```

```python
df["category"] = df["category"].str.title()
```

```python
df["month"] = df["date"].dt.month
df["year"] = df["date"].dt.year
df["day_of_week"] = df["date"].dt.day_name()
df['day'] = df['date'].day
```

```python
df["category"] = df["category"].str.strip().str.lower()
```

```python
category_map = {
    "Transport": "Transport",
    "Rent Car": "Transport",
    "Taxi": "Transport",
    "Travel": "Transport",
    "Fuel": "Transport",
    "Restuarant" : "Food",
    "Coffe":"Food",
    'Business Lunch' :"Food"
}

df["category"] = df["category"].replace(category_map)
```

```python
Monthly_expense = df.groupby(['year','month','category']).sum('amount')
```

```python
display(Monthly_expense)
```

```python
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["year_month"] = df["date"].dt.to_period("M").astype(str)  # e.g. "2025-08"
df["week"] = df["date"].dt.isocalendar().week
df["dow"] = df["date"].dt.day_name()
```

```python
df.head()
```

```python
Monthly_expense = (
    df.groupby(["year_month"])["amount"]
      .sum()
      .reset_index()   # <-- makes year_month a column
)
```

```python
display(Monthly_expense)
```

```python
import matplotlib.pyplot as plt
```

```python
Monthly_expense.plot(x='year_month',y='amount',kind='bar')
```

```python
Monthly_category_expense = (
    df.groupby(["category"])["amount"]
      .sum()
      .reset_index()   # <-- makes year_month a column
)

Monthly_category_expense.plot(x='category',y='amount',kind='bar')
```
